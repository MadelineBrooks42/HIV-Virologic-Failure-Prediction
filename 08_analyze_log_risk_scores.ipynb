# Distribution of log risk scores and simulated-data coefficient summaries

**Notebook:** `08_analyze_log_risk_scores.ipynb`

## Purpose

- Reads the log risk scores written by scripts 01-06 and plots histograms of log risk scores and risk scores (Cox vs DeepSurv) for each feature-selection method.
- Repeats the log risk score plots for runs on simulated data, and prints mean / SD of feature coefficients from simulated time-fixed and time-varying runs.

## Inputs (read from `INPUT_DIR`)

- `cox_feature_subset_LogRiskScores_ByFold.csv`
- `cox_powershap_LogRiskScores_ByFold.csv`
- `cox_shap_filter_LogRiskScores_ByFold.csv`
- `deepsurv_feature_subset_LogRiskScores.csv`
- `deepsurv_powershap_LogRiskScores.csv`
- `deepsurv_shap_filter_LogRiskScores.csv`
- Results of runs of scripts 01-06 on simulated data, stored in `simulated/`, `simulated/time_fixed/` and `simulated/time_varying/` under `INPUT_DIR`. Scripts 01-06 use time-fixed covariates, so the `time_varying` files cannot be produced by them (see the README)

## Outputs (written to the figures folder)

- `log_risk_score_distribution_feature_subset.png`
- `log_risk_score_distribution_shap.png`
- `log_risk_score_distribution_powershap.png`

## Configuration

| Variable | Value in this notebook | Meaning |
|---|---|---|
| `INPUT_DIR` | `"output/"` | Folder holding the CSVs from scripts 01-06 (must end with `/`) |
| `OUTPUT` | `"figures"` | Folder where figures are saved (must exist) |

## Requirements

Python packages imported by this notebook: `matplotlib`, `numpy`, `pandas`.


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Folder containing the CSV files written by scripts 01-06 (must end with "/")
INPUT_DIR = "output/"
# Folder where figures are saved (must exist before running)
OUTPUT = "figures"
```

# Feature Subset

```python
cox = pd.read_csv(f"{INPUT_DIR}cox_feature_subset_LogRiskScores_ByFold.csv")
deepsurv = pd.read_csv(f"{INPUT_DIR}deepsurv_feature_subset_LogRiskScores.csv")

# Convert log risk scores by exponentiating them (base e) to get risk scores
cox["risk"] = np.exp(cox["log_risk"])
deepsurv["risk"] = np.exp(deepsurv["log_risk"])
```

```python
# plot histograms of log risk scores for both models on the same plot
plt.figure(figsize=(10, 6))
plt.hist(cox["log_risk"], bins=30, alpha=0.5, label="Cox")
plt.hist(deepsurv["log_risk"], bins=30, alpha=0.5, label="DeepSurv")
plt.xlabel("Log Risk Score")
plt.ylabel("Frequency")
plt.title("Distribution of Log Risk Scores for Cox and DeepSurv Models for Science-Driven Feature Selection")
plt.legend()

# add vertical lines at the mean log risk score for each model
plt.axvline(x=cox["log_risk"].mean(), color="red", linestyle="--", label="Cox Mean Log Risk Score")
plt.axvline(x=deepsurv["log_risk"].mean(), color="blue", linestyle="--", label="DeepSurv Mean Log Risk Score")

# add a legend for the vertical lines
plt.legend()

# save the plot to the output directory
plt.savefig(f"{OUTPUT}/log_risk_score_distribution_feature_subset.png")
```

```python
# plot histograms of log risk scores for both models on the same plot
plt.figure(figsize=(10, 6))
plt.hist(cox["risk"], bins=30, alpha=0.5, label="Cox")
plt.hist(deepsurv["risk"], bins=30, alpha=0.5, label="DeepSurv")
plt.xlabel("Risk Score")
plt.ylabel("Frequency")
plt.title("Distribution of Risk Scores for Cox and DeepSurv Models")
plt.legend()

# add a vertical line at x=1
plt.axvline(x=1, color="black", linestyle="--", label="Risk Score = 1")

# add vertical lines at the mean risk score for each model
plt.axvline(x=np.exp(cox["log_risk"].mean()), color="red", linestyle="--", label="Cox Mean Risk Score Exponentiated")
plt.axvline(x=np.exp(deepsurv["log_risk"].mean()), color="blue", linestyle="--", label="DeepSurv Mean Risk Score Exponentiated")

# add a legend for the vertical lines
plt.legend()
```

## Time Fixed Coefficients

### Cox

```python
import pandas as pd
```

```python
df = pd.read_csv(f"{INPUT_DIR}simulated/time_fixed/cox_feature_subset_FeatureCoefficients.csv")

# Calculate the mean coefficient for each feature
mean_coefficients = df.groupby('feature')['coefficient'].mean().reset_index()

# calculate the 95% standard deviation for each feature
std_coefficients = df.groupby('feature')['coefficient'].std().reset_index()

print(mean_coefficients)
print("")
print(std_coefficients)
```
